# Eksplorasi Data (EDA)
Notebook ini bertujuan untuk melakukan visualisasi awal pada dataset untuk melihat sampel dan intensitas warna.

```python
import sys
sys.path.append('..') # Agar bisa import module dari folder src dan config

import cv2
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from config import *
from src.preprocess import download_data_from_kaggle

kaggle_dir = download_data_from_kaggle()
print("Data raw berada di:", kaggle_dir)
```

## 1. Visualisasi Sampel Kelas

```python
def show_samples(split='Train'):
    plt.figure(figsize=(10, 5))
    for i, cls in enumerate(CLASSES):
        img_paths = list((kaggle_dir / split / cls).glob("*.png")) + list((kaggle_dir / split / cls).glob("*.jpg"))
        if len(img_paths) > 0:
            img = cv2.imread(str(img_paths[0]))
            img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            plt.subplot(1, 2, i+1)
            plt.imshow(img_rgb)
            plt.title(f"Sample: {cls}")
            plt.axis('off')
    plt.show()

show_samples()
```

## 2. Visualisasi Distribusi Warna (Grayscale, HSV, LAB)

```python
def plot_color_spaces(img_path):
    img = cv2.imread(str(img_path))
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    img_lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
    
    fig, axes = plt.subplots(2, 4, figsize=(16, 8))
    
    # Gambar
    axes[0, 0].imshow(img_rgb)
    axes[0, 0].set_title("Original RGB")
    axes[0, 1].imshow(img_gray, cmap='gray')
    axes[0, 1].set_title("Grayscale")
    axes[0, 2].imshow(img_hsv)
    axes[0, 2].set_title("HSV")
    axes[0, 3].imshow(img_lab)
    axes[0, 3].set_title("LAB")
    
    # Histogram
    axes[1, 0].hist(img_rgb.ravel(), bins=50, color='blue', alpha=0.7)
    axes[1, 0].set_title("RGB Histogram")
    axes[1, 1].hist(img_gray.ravel(), bins=50, color='gray', alpha=0.7)
    axes[1, 1].set_title("Grayscale Histogram")
    axes[1, 2].hist(img_hsv[:,:,0].ravel(), bins=50, color='orange', alpha=0.7)
    axes[1, 2].set_title("Hue Histogram")
    axes[1, 3].hist(img_lab[:,:,0].ravel(), bins=50, color='green', alpha=0.7)
    axes[1, 3].set_title("Lightness Histogram")
    
    plt.tight_layout()
    plt.show()

# Ambil sampel pertama dari kelas WithMask
sample_path = list((kaggle_dir / 'Train' / CLASSES[0]).glob("*.png"))[0]
plot_color_spaces(sample_path)
```

Notebook ini digunakan sebagai arena bermain (*sandbox*) untuk bereksperimen dengan kombinasi hyperparameter SVM atau layer baru pada CNN tanpa merusak pipeline produksi (`kaggle_train.ipynb`).

```python
import sys
sys.path.append('..')

from config import *
from src.preprocess import download_data_from_kaggle
from src.feature_engineering import load_and_extract_features, scale_features
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

kaggle_dir = download_data_from_kaggle()
```

## Uji Coba Kernel SVM Lain (Contoh: Linear atau Poly)

```python
# Mari ambil sedikit sampel dari Validation untuk eksperimen cepat
X_val_c, X_val_d, y_val = load_and_extract_features(kaggle_dir / "Validation", require_preprocess=True)
X_test_c, X_test_d, y_test = load_and_extract_features(kaggle_dir / "Test", require_preprocess=True)

X_val_c_s, X_test_c_s, _, _ = scale_features(X_val_c, X_test_c, None)

print("Mencoba SVM Linear...")
svm_linear = SVC(kernel='linear')
svm_linear.fit(X_val_c_s, y_val)
pred = svm_linear.predict(X_test_c_s)
print("Akurasi eksperimen SVM Linear pada Test Set:", accuracy_score(y_test, pred))
```
